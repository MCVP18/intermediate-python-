```python
def count_lines(filename):
    try:
        file_obj = open(filename)
        return len(file_obj.readlines())
    finally:
        file_obj.close()
#---------------------------------------
count_lines('fruits.txt')
```

#### Using with statement
- We don't have to close the resource explicitly
- Resources get closed after execution of the 'with' block
- 'with' block is an alternate to try-finally pattern

```python
def count_lines(filename):
    with open(filename) as file_obj:
        return len(file_obj.readlines()) 
#---------------------------------------
count_lines('fruits.txt')
```

```python
class Something:
    def __init__(self):
        print('Object created')
    
    def do_something(self):
        print('Object doing a task')
        
    def close(self):
        print('Close called')
```

```python
try:
    obj = Something()
    obj.do_something()
finally:
    obj.close()
```

```python
from contextlib import closing

with closing(Something()) as obj:
    obj.do_something()
```
