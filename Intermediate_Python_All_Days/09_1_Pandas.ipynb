#### Pandas Series
- Customizable arrays

```python
labels = ['a','b','c','d']
list1 = [10, 20, 30, 40]
arr = np.array(list1)
dict1 = {'a':10, 'b':20, 'c':20, 'd':40}
```

```python
series = pd.Series(list1)
#print(type(series))
series
```

```python
series[2]
```

```python
series = pd.Series(data=list1, index=labels)
series
```

```python
series['c']
```

```python
pd.Series(arr) # Series created from Numpy array
```

```python
pd.Series(arr, index=labels)
```

```python
pd.Series(dict1)
```

```python
days = ['Monday','Tuesday','Wednesday','Thursday','Friday']
labels = [1,2,3,4,5]
series = pd.Series(days, index=labels)
series
```

```python
series[3]
```

```python
groceries = {'apple': 6, 'banana':12, 'orange':10, 'pineapple':2, 'pear':4}
series = pd.Series(groceries)
series
```

```python
series['apple']
```

```python
series[1:4]
```

```python
groceries2 = {'apple': 3, 'banana':5, 'orange':7}
series2 = pd.Series(groceries2)
series2
```

```python
groceries_series = series + series2
groceries_series
```

```python
groceries_series.isnull()
```

```python
groceries_series.notnull()
```

```python
groceries_series
```

```python
groceries_series = groceries_series.fillna(0)
groceries_series
```

#### Missing Data

```python
df = pd.DataFrame(
        {
        'A' : [1, 2, np.nan],
        'B' : [4, np.nan, np.nan],
        'C' : [7, 8, 9]   
        }
    )

df
```

```python
df.dropna() # Drop rows which have atleast one missing data
```

```python
df.dropna(axis=1) # Drop cols which have atleast one missing data
```

```python
df
```

```python
df.dropna(thresh=2) # Retain rows which have atleast 2 non null columns
```

```python
df
```

```python
df.dropna(thresh=2 ,axis=1) # Retain columns which have atleast 2 non null columns
```

```python
df
```

```python
df.fillna(0)
```

```python
df.fillna(df.mean())
```

```python
df
```

```python
df['A'].fillna(value=df['A'].mean()) # Fetch one column only
```
