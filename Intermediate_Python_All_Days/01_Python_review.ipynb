```python
s = input('Enter month (1-12): ')

if s.isnumeric():
    month = int(s)
    
    match month:
        case 1 | 2 | 3:
            print('1st Quarter')
        case 4 | 5 | 6:
            print('2nd Quarter')
        case 7 | 8 | 9:
            print('3rd Quarter')
        case 10 | 11 | 12:
            print('4th Quarter')
        case _:
            print('Please enter a value between 1 and 12')
else:
    print('Please enter number only')
```

#### List assignment
- get the count of an element (number of occurences)
- get all the indices where the element occurs
- Part B --> Redo the assignment using list comprehension and enumerate feature

```python
fruits = ['apple', 'banana', 'orange', 'pineapple', 'banana']
fruits_to_find = 'banana'
```

```python
count = fruits.count(fruits_to_find)

if count > 0:
    indices = []
    
    for i in range(0,len(fruits)):
        if fruits[i] == fruits_to_find:
            indices.append(i)
    print(f'{fruits_to_find} occurs {count} times(s)')
    print(f'{fruits_to_find} occurs at {indices}')
    
else:
    print(fruits_to_find, 'does not exist')
```

```python
count = fruits.count(fruits_to_find)

if count > 0:
    
    indices = [i for i,ele in enumerate(fruits) if ele == fruits_to_find]
    
    print(f'{fruits_to_find} occurs {count} times(s)')
    print(f'{fruits_to_find} occurs at {indices}')
else:
    print(fruits_to_find, 'does not exist')
```

#### Flatten a Tuple

```python
tuple1 = (1, 2, 3, (4,5,6))
print(tuple1)

temp_list = []

for ele in tuple1:
    if type(ele) is tuple:
        temp_list.extend(ele)
    else:
        temp_list.append(ele)

tuple1 = tuple(temp_list) #   

print(tuple1)
```

#### Write a function which takes a string as parameter and displays the words and their frequency (ignore periods)
1. Case-sensitive
2. Case-insensitive

```python
def word_frequency(string1):
    
    #case-sensitive
    #word_list = string1.replace('.','').split()
    
    #case-insensitive
    word_list = string1.replace('.','').lower().split()
    
    #get unique words
    unique_words = set(word_list)
    
    for word in unique_words:
        print(f'{word} = {word_list.count(word)}')
    
#------------------------------------

s1 = 'Hello. I Like python. Python is a simple language. Hello world from Python.'
word_frequency(s1)
```
