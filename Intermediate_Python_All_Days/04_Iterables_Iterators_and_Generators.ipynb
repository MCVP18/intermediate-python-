#### Container
- Data structure which holds elements
- Supports the membership tests (it can be asked whether it contains a certain element)
- Operators are 'in' and 'not in' 
- Examples --> string, list, set, tuple, dict

```python
'L' not in 'python'
```

```python
5 in [1,2,3,4,5]
```

```python
'Like' in ('I', 'Like', 'Python')
```

```python
'Name' in {'Name':'John', 'Age':23}
```

```python
'John' in {'Name':'John', 'Age':23}.values()
```

#### Iterables
- Any object that can return an Iterator
- The method \_\_iter\_\_() returns the Iterator
- Any object which has \_\_iter\_\_() if Iterable
- Most containers are Iterables
- Files, Sockets are iterables as well

```python
list1 = [1,2,3]
iter1 = list1.__iter__()

print(type(iter1))
```

```python
iter2 = iter(list1)
print(type(iter2))
```

#### Iterator
- A stateful object that produces the next value when we call \_\_next\_\_() method on it OR next(iterator)
- Any object that has \_\_next\_\_() method is an Iterator
- How does it produce the next value is irrelevant
- Iterator is a value factory

```python
list1 = [1,2,3]
iter1 = list1.__iter__()
```

```python
iter1.__next__()
```

```python
iter2 = iter(list1)
```

```python
next(iter2)
```

```python
# Internally, for loop uses the iterator provided by the iterable object
for ele in list1:
    print(ele)
```

#### using the while loop

```python
iter1 = list1.__iter__()

try:
    while True:
        print(iter1.__next__())
except StopIteration:    
    pass
```

```python
iter2 = iter(list1)

try:
    while True:
        print(next(iter2))
except StopIteration:
    pass
```

#### Let's build our own Iterator
#### Generate Fibonacci sequence till the value is less than the specified passed as constructor parameter

```python
class Fibonacci:
    def __init__(self, max):
        self.max = max
        self.curr = 0
        self.next = 1
        
    def __iter__(self):
        return self
    
    def __next__(self): # stop generating when value exceeds max
        value = self.curr
        self.curr = self.next
        self.next = value + self.curr
        
        if value > self.max:
            raise StopIteration
        
        return value
#-------------------------------------------

f = Fibonacci(2000)

for ele in f:
    print(ele, end='  ')    
```

#### Generators
- A Generator allows us to write code which behaves like an Iterator without having methods like \_\_iter\_\_() and \_\_next\_\_()
- Every Generator is an Iterator (but not vice versa)
- Two types of Generators
    - Generator functions
    - Generator expressions

```python
import keyword

print(keyword.kwlist)
```

```python
def simple_generator():
    yield 1
    yield 'Python'
    yield 3.1415
#----------------------------

for value in simple_generator():
    print(value)
```

```python
# Generator function which generates odd numbers (how_many) from thes start:
def odd_generator(start, how_many):
    count = 0
    while count < how_many:
        if start % 2 != 0:
            count += 1
            yield start
        start += 1
#-----------------------------

for value in odd_generator(6,10):
    print(value)
```

#### Write a generator function which reverses a string (yield one char at a time in reverse order)

```python
def string_reverse(string1):
    for i in range(len(string1)-1, -1, -1):
        yield(string1[i])
#-----------------------------

for ch in string_reverse('civic'):
    print(ch, end='')
```
