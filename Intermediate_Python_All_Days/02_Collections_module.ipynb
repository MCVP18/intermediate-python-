#### Default Dictionary 

```python
from collections import defaultdict
```

```python
dd = defaultdict(int) # int --> 0, str --> '', list --> []

dd['a'] = 'Apple'
dd['z'] = 'Zebra'

print(dd['a'])
print(dd['z'])
print(dd['d'])
```

```python
def default_value():
    return 'Not set'

dd = defaultdict(default_value) # function gets called in case the key does not exist

dd['a'] = 'Apple'
dd['z'] = 'Zebra'

print(dd['a'])
print(dd['z'])
print(dd['d'])
```

```python
s = 'Helloworld'

letters = {}

for ch in s:
    letters[ch] = letters.get(ch,0) + 1

letters
```

```python
s = 'Helloworld'

letters = defaultdict(int)

for ch in s:
    letters[ch] = letters[ch] + 1

letters
```

#### Default Dictionary assignment
Read from a file (fruits.txt) where each line contains a fruit name followed by a count
- apple 2
- banana 5
- pear 3
- banana 7
- apple 3
- cherry 100
- apple 5
- pear 6

Part-A --> {'apple':[2,3,5], 'banana':[5,7], 'pear':[3,6], 'cherry':[100]}

Part-B --> {'apple':10, 'banana':12, 'pear':9, 'cherry':100}

#### Part A

```python
fruits = defaultdict(list)

with open('fruits.txt') as fruits_file:
    line = fruits_file.readline()
    while line:
        fruit,count = line.split()
        fruits[fruit].append(int(count))
        line = fruits_file.readline()

fruits        
```

#### Part B

```python
fruits = defaultdict(int)

with open('fruits.txt') as fruits_file:
    line = fruits_file.readline()
    while line:
        fruit,count = line.split()
        fruits[fruit] = fruits[fruit] + int(count)
        line = fruits_file.readline()

fruits   
```

#### Named Tuple

```python
from collections import namedtuple
```

```python
Point = namedtuple('Point', ['x','y'])

p = Point(2,5)

print(type(p))

print(p[0])
print(p[1])
```

```python
print(p.x)
print(p.y)
```

```python
p.x = 100 # Error since tuple is Immutable
```

```python
Employee = namedtuple('Employee', ['id','name','designation'])

e1 = Employee(101, 'John', 'Programmer')
e1
```

```python
print(f'Id: {e1.id}')
print(f'Name: {e1.name}')
print(f'Designation: {e1.designation}')
```

```python
# create named tuple instance from values present in a list
list1 = [102, 'Sam', 'Admin']

e2 = Employee._make(list1)
e2
```

```python
# get field names of a named tuple
Employee._fields
```

```python
# create named tuple instance from key/value pairs present in a dict
dict1 = {'id':103, 'name':'Mat', 'designation':'Accountant'}

e3 = Employee(**dict1)
e3
```

```python
e3 = e3._replace(name='Mathew') # create and return a new object
e3
```
