#### Python scoping rules
- LEGB (Local, Enclosed, Global, Built-in)
- Local: Declared within a function
- Enclosed: Declared in the outer function
- Global: Declared at the uppermost later (within a script)
- Built-in: Declarations which are part of Core Library

```python
a = '10'

def outer_func():
    b = 100
    def inner_func():
        #b = 20
        c = 30
        print(c) # Local
        print(b) # Enclosed
        print(a) # Global
        
    inner_func()
```

```python
outer_func()
```

#### Closure
A Closure is an inner function which remembers the value(s) from the enclosing scope even when the flow is no longer on the enclosing scope

```python
def make_adder(x):
    
    def adder(y):
        return x + y
    
    return adder
```

```python
inner_func1 = make_adder(10)
inner_func1(20) # x is 10
```

```python
inner_func2 = make_adder(100)
inner_func2(20) # x is 100
```

#### Decorator
- Wrapper functions, built on top of Closure feature
- Outer function has a parameter which is a function to be decorated (target function)
- Inner function accepts same parameters as that of target function
- Purpose of Decorator is to modify the function behavior without modifying the function code
- Used to adding Pre or Post or both actions for a function

#### Closure vs Decorator
- Decorator should accept target function as a parameter (No such requirement in a Closure)
- Inner function should acceptsame parameters as that of target function (No such requirement in a Closure)
- All Decorators are Closures but vice-versa may not be true 

```python
def isPositive(target_func):
    def inner_func(num):
        if num > 0:
            print(f'Calling target function --> {target_func.__name__}')
            result = target_func(num)
            return result
        else:
            print('WARNING: Negative Number not accepted')
    return inner_func
```

```python
@isPositive
def demo(num):
    print(f'Number is {num}')
```

```python
demo(100)
```

```python
class Person:
    def __init__(self, age):
        self.__age = age
    
    @property
    def age(self):
        return self.__age
    
    @age.setter
    def age(self,value):
        if value >= 0 and value <= 99:
            self.__age = value
        else:
            print(f'Invalid value {value} for age')
#----------------------------------------------

p = Person(23)
print(p.age) # access as if age is a variable
p.age = 99 # instead of p.set_age(99)
print(p.age)
```
