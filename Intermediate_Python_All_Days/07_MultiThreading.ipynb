#### Create a new Thread (approach 1)
1. Write a function which will be executed by the Thead when it is scheduled
2. Create instance of Thread class and associate the target function with it
3. Invoke start() method on the Thread instance

```python
def run():
    for i in range(1,6):
        print(f'{i} - Thread name: {threading.current_thread().name}, Process id: {os.getpid()}')
        time.sleep(0.5)
```

```python
print(f'Main Thread enter, Process id: {os.getpid()}')
t1 = threading.Thread(target=run, name='FirstThread')
t1.start()
time.sleep(5) # <---- current thread goes to sleep state ---->
print('Main Thread exit')
```

```python
print(f'Main Thread enter, Process id: {os.getpid()}')

t1 = threading.Thread(target=run, name='FirstThread')
t1.start()
print(f'\tt1 is alive - {t1.is_alive()}') # Maybe True or False
t1.join()  # <---- current thread (main thread) goes to wait state, gets notified when t1 is dead  ---->
print(f'\tt1 is alive - {t1.is_alive()}') # Definitely False
print('Main Thread exit')
```

```python
print(f'Main Thread enter, Process id: {os.getpid()}')

t1 = threading.Thread(target=run, name='Thread-01')
t2 = threading.Thread(target=run, name='Thread-02')

t1.start()
t2.start()

t1.join()
t2.join() 

print('Main Thread exit')
```

#### Create Thread by Inheritence
- Write a sub-class of Thread and instantiate the sub-class
- Override the run method (it is the target method by default)
- arguments can be set explicitly through 'args' attribute and accessed using self.args

```python
class MyThread(threading.Thread):
    def run(self):
        for i in range(1,6):
            print(f'{i} - Thread name: {threading.current_thread().name}, Process id: {os.getpid()}')
            time.sleep(0.5)

#------------------------------------------

t1 = MyThread()
t1.start()
```

```python
class MyThread(threading.Thread):
    def do_something(self):
        for i in range(1,6):
            print(f'{i} - Thread name: {threading.current_thread().name}, Process id: {os.getpid()}')
            time.sleep(0.5)

#------------------------------------------

t1 = MyThread()
t1.name = 'First'
t1.run = t1.do_something # explicitly set the target since the method is not named as 'run'

t2 = MyThread()
t2.name = 'Second'
t2.run = t2.do_something # explicitly set the target since the method is not named as 'run'

t1.start()
t2.start()
```

#### Write a sub-class of Thread named AddThread
- the run method sums the arguments and display the sum
- arguments can be set explicitly through 'args' attribute and accessed using self.args

```python
class AddThread(threading.Thread):
    def run(self):
        print(f'{threading.current_thread().name} --> {sum(self.args)}')
```

```python
t1 = AddThread()
t1.name = 'First'
t1.args = (10,20,30)

t2 = AddThread()
t2.name = 'Second'
t2.args = (100,200,300)

t1.start()
t2.start()
```

#### Thread assignment
1. Write a class named Resource
2. It would have an instance field named 'data'
3. It would have an instance method namaed 'do_something'
4. The method would increment data by 1 and display the name of the current thread and value of the data
5. The resource method 'do_something' will be the target method for a Thread

```python
class Resource:
    def __init__(self):
        self.data = 0
        
    def do_something(self):
        self.data += 1
        time.sleep(0.01) # Simulate context switch
        print(f'{threading.current_thread().name} - {self.data}')
```

```python
class Resource:
    def __init__(self):
        self.data = 0
        self.lock = threading.Lock()
        
    def do_something(self):
        self.lock.acquire()
        self.data += 1
        time.sleep(0.01) # Simulate context switch
        print(f'{threading.current_thread().name} - {self.data}')
        self.lock.release()
```

```python
r = Resource()

t1 = threading.Thread(target = r.do_something, name='First')
t2 = threading.Thread(target = r.do_something, name='Second')
t3 = threading.Thread(target = r.do_something, name='Third')

t1.start()
t2.start()
t3.start()
```

#### MultiProcessing in Python allows us to run tasks in parallel
- MultiThreading - Concurrent processing (uses only 1 CPU which is shared by all the Threads of the process)
- MultiProcessing - Parallel processing (uses as many CPU as availble)

```python
import os
import multiprocessing
```

```python
print(f'Available Processors: {os.cpu_count()}')
```

```python
def worker1():
    print(f'In Worker-1, Process id: {os.getpid()}')
    
def worker2():
    print(f'In Worker-2, Process id: {os.getpid()}')

print(f'Main Process id: {os.getpid()}')

p1 = multiprocessing.Process(target=worker1)
p2 = multiprocessing.Process(target=worker2)

p1.start()
p2.start()
```

#### Create 2 Process objects
- One will square a given number(as parameter)
- Other will cube a given number(as parameter)
- Parameters are passed as 'args' (similar to the args concept in threading)
- Make sure that only one process prints to the standard output at a time(use a common process lock) 

```python
def square(lock,num):
    lock.acquire()
    print(f'Square process id: {os.getpid()}')
    print(f'Square of {num} is {num ** 2}')
    lock.release()
    
def cube(lock,num):
    with lock:
        print(f'Cube process id: {os.getpid()}')
        print(f'Cube of {num} is {num ** 3}')

print(f'Main Process id: {os.getpid()}')

lock = multiprocessing.Lock()
p1 = multiprocessing.Process(target=square, args=(lock,10))
p2 = multiprocessing.Process(target=cube, args=(lock,10))

p1.start()
p2.start()

p1.join()
p2.join()

print('Main Process exit')
```
